```python
np.random.seed(42)
marks = np.random.randint(20, 101, (12, 4))
averages = marks.mean(axis=1)
test_means = marks.mean(axis=0)
test_std = marks.std(axis=0)
passed = np.sum(np.all(marks >= 40, axis=1))
high_students = np.where(averages >= 75)[0] + 1
most_variable_test = np.argmax(test_std) + 1
print("Marks:\n", marks)
print("Student averages:", averages)
print("Test means:", test_means)
print("Test standard deviations:", test_std)
print("Students passing all tests:", passed)
print("Students averaging >= 75:", high_students)
print("Test with greatest variation:", most_variable_test)
```

```python
a = np.arange(21, 46).reshape(5, 5)
print("Original matrix:\n", a)
central = a[1:4, 1:4]
print("Central matrix:\n", central)
transposed = central.T
print("Transpose:\n", transposed)
product = central @ transposed
print("Product:\n", product)
row_sums = product.sum(axis=1)
print("Row sums:", row_sums)
maximum = product.max()
positions = np.argwhere(product == maximum)
print("Maximum:", maximum)
print("Maximum positions:", positions)
```

```python
quantity = np.array([
    [10, 12, 15, 8],
    [20, 18, 12, 10],
    [15, 20, 18, 12],
    [25, 15, 20, 10],
    [18, 22, 15, 15],
    [20, 18, 25, 12]
])
original_prices = np.array([150, 240, 90, 320])
daily_revenue = quantity @ original_prices
product_quantity = quantity.sum(axis=0)
product_revenue = product_quantity * original_prices
print("Daily revenue:", daily_revenue)
print("Product quantities:", product_quantity)
print("Revenue per product:", product_revenue)
print("Highest revenue day:", np.argmax(daily_revenue) + 1)
print("Best-selling product:", np.argmax(product_quantity) + 1)
new_prices = np.where(product_quantity > 100,
                      original_prices * 0.9,
                      original_prices)
new_revenue = quantity @ new_prices
print("Original prices:", original_prices)
print("Reduced prices:", new_prices)
print("New total revenue:", new_revenue.sum())
```

```python
np.random.seed(42)
data = np.random.normal(50, 12, 100)
cleaned = np.where(data < 30, 30,
                   np.where(data > 70, 70, data))
print("Values outside 30-70:", np.sum((data < 30) | (data > 70)))
print("\nOriginal statistics:")
print("Mean:", data.mean())
print("Variance:", data.var())
print("Standard deviation:", data.std())
print("Minimum:", data.min())
print("Maximum:", data.max())
print("\nCleaned statistics:")
print("Mean:", cleaned.mean())
print("Variance:", cleaned.var())
print("Standard deviation:", cleaned.std())
print("Minimum:", cleaned.min())
print("Maximum:", cleaned.max())
```

```python
branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})
result = branch_a + branch_b + branch_c
print("Ordinary addition:")
print(result)
common = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
print("Common quarters:", common.tolist())
combined = branch_a[common] + branch_b[common] + branch_c[common]
print("Combined earnings for common quarters:")
print(combined)
zero_result = (
    branch_a.add(branch_b, fill_value=0)
            .add(branch_c, fill_value=0)
)
print("Addition treating absent records as zero:")
print(zero_result)
```

```python
scores = {
    'R101': 78,
    'R102': 65,
    'R104': 92,
    'R105': 48
}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']
s = pd.Series(scores, index=required)
print("Original Series:")
print(s)
print("\nMissing records:")
print(s[s.isnull()])
print("\nValid scores >= 70:")
filtered = s[s.notnull() & (s >= 70)]
print(filtered)
d = filtered.to_dict()
print("\nDictionary:")
print(d)
reconstructed = pd.Series(d)
print("\nReconstructed Series:")
print(reconstructed)
print("\nValues match:", filtered.equals(reconstructed))
```
