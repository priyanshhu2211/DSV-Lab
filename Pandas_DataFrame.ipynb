```python
product_df = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet'],
    'Category': ['Electronics', 'Electronics', 'Electronics'],
    'Price': [55000, 25000, 18000]
})
product_ordered_df = product_df[['Price', 'Product', 'Category']]
product_ordered_df
```

```python
course_df = pd.DataFrame({
    'Student': ['Aarav', 'Diya', 'Kabir', 'Meera', 'Rohan'],
    'Course': ['Python', 'Java', 'C', 'Python', 'Java'],
    'Score': [88, 91, 76, 95, 84]
})
print(course_df.columns)
display(course_df['Student'])
student_score_df = course_df[['Student', 'Score']]
display(student_score_df)
course_df['Batch'] = '2026'
course_df
```

```python
travel_df = pd.DataFrame({
    'Destination': ['Agra', 'Jaipur', 'Delhi', 'Lucknow', 'Varanasi']
})
travel_df['Code'] = np.arange(5)
display(travel_df)

travel_codes = pd.Series([101, 104], index=[1, 4])
travel_df['Code'] = travel_codes
display(travel_df)
```

```python
school_df = pd.DataFrame({
    'Student': ['Anaya', 'Vihaan', 'Sara', 'Arjun'],
    'Class': [9, 10, 9, 10],
    'Section': ['A', 'B', 'A', 'C']
})
school_extended_df = school_df.reindex(columns=['Student', 'Class', 'Section', 'House'])
school_extended_df['House'] = 'Blue House'
school_extended_df
```
